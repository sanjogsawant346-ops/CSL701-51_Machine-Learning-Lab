```python
#Task 1
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Task 2: Load the Titanic dataset
import pandas as pd
df = pd.read_csv("/content/Titanic-Dataset (4).csv")
print(df.head())
```

```python
# Task 3: Display the first 10 rows of the dataset

df.head(10)
```

```python
# Task 4: Display the last 5 rows of the dataset

df.tail(5)
```

```python
# Task 5: Check the dimensions of the dataset

print("Dataset Shape:", df.shape)
```

```python
# Task 6: Display dataset information

df.info()
```

```python
#Task 7: Generate summary statistics using df.describe(); record mean, minimum and
print(df.describe())
print("\nAge Statistics:")
print("Mean Age:", df["Age"].mean())
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())

print("\nFare Statistics:")
print("Mean Fare:", df["Fare"].mean())
print("Minimum Fare:", df["Fare"].min())
print("Maximum Fare:", df["Fare"].max())
```

```python
#Task 8: Identify missing values using df.isnull().sum().
print(df.isnull().sum())

```

```python
#Task 9: Fill missing Age values with the median.
median_age = df["Age"].median()

df["Age"] = df["Age"].fillna(median_age)

print("Missing values in Age after filling:", df["Age"].isnull().sum())
```

```python
#Task 10: Filter surviving passengers (Survived == 1).

survivors = df[df["Survived"] == 1]
print(survivors.head(10))
print("\nTotal Surviving Passengers:", survivors.shape[0])
```

```python
#Task 11: Convert Fare column to a NumPy array.
fare_array = df["Fare"].to_numpy()
print(fare_array)
print("\nType:", type(fare_array))
```

```python
#Task 12: Calculate maximum and average Fare using np.max() and np.mean().
import numpy as np
max_fare = np.max(fare_array)
mean_fare = np.mean(fare_array)
print("Maximum Fare:", max_fare)
print("Average Fare:", mean_fare)
```

```python
#Task 13: Plot histogram of Age.

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df["Age"], bins=20, edgecolor="black")

plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
#Task 14: Plot bar chart of Survival Count.
import matplotlib.pyplot as plt
survival_count = df["Survived"].value_counts()
plt.figure(figsize=(6, 5))
plt.bar(["Not Survived", "Survived"], survival_count, color=["red", "green"])
plt.title("Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
#Task 15: Plot scatter graph of Age vs Fare.
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.scatter(df["Age"], df["Fare"], alpha=0.6)
plt.title("Scatter Plot of Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```
